```python
%sql
ALTER TABLE workspace.default.databrics_data
RENAME TO workspace.default.orders_2025;
```

```python
%sql
-- Question B. How many orders were delivered and how many were cancelled?
SELECT order_status, COUNT(*) as orders
FROM orders_2025
GROUP BY order_status;
```

```python
%sql
-- Questoin c. How many orders were placed in each month of 2025
SELECT month(order_purchase_timestamp) As month, COUNT(*) as orders
From orders_2025
GROUP BY month 
ORDER BY month
```

```python
%sql
 -- E1. What is the average number of days from purchase to delivery for delivered orders?
SELECT
    AVG(DATEDIFF(order_delivered_customer_date, order_purchase_timestamp)) AS avg_delivery_days
FROM orders_2025
WHERE order_status = 'delivered';
```

```python
%sql
-- E2. On which single day in 2025 were the most orders placed?
SELECT
    DATE(order_purchase_timestamp) AS order_date,
    COUNT(*) AS order_count
FROM orders_2025
GROUP BY DATE(order_purchase_timestamp)
ORDER BY order_count DESC
LIMIT 1;

```

```python
%sql
-- E3. How many delivered orders arrived after their estimated delivery date?
SELECT
    COUNT(*) AS late_delivered_orders
FROM orders_2025
WHERE order_status = 'delivered'
  AND order_delivered_customer_date > order_estimated_delivery_date;
```

```python
%sql
--E4. Which month has the highest cancellation rate, and why might that be?
SELECT
    MONTH(order_purchase_timestamp) AS month,
    AVG(CASE
        WHEN order_status = 'canceled' THEN 1.0
        ELSE 0.0
    END) AS cancellation_rate
FROM orders_2025
GROUP BY MONTH(order_purchase_timestamp)
ORDER BY cancellation_rate DESC
LIMIT 1;
```
